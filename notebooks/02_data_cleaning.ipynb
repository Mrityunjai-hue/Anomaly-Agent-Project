## Missing Delivery Date Analysis

Missing Delivery dates are expected for orders that have not reached the delivered stage.

however 8 orders have 'order_status = delivered' while 
'order_delivered_customer_date' is missing, these records are retained because the true delivery date is unknown but theu should be treated as data-quality issue exceptions.



```python
orders[orders['order_approved_at'].isna()][['order_status', 'order_purchase_timestamp', 'order_approved_at']].head(20)
```

### missing approval timestamp analysis

most missing 'order_apprved_at' values occur for cancelled or created orders, where an approval timestamp may legitimately not exist.

a small nnumber of delivered orders also have a missing approval timestamp, these records are retained because the true approval time is unknown, but they should be treated as data-quality exceptions.

### Orders Data Cleaning Decision 

datetime columns converted to proper datetime types

missing delivery and approval timestamps looked on basis of order_status. missing timestamps are retained as NaT because they can represent legitimate stages of the order lifecycle.

data quality exceptions
- delivered orders with missing customer delivery date
- delivered orders with missing approval timestamp

these exceptions will be retained and handled as data-quality signals rather than being filled with artificial values
